```python
def fibonacci_no(position):
    old = new = 1
    for i in range(position-1):
        new, old = old, new+old
    return new
```

```python
def rabbits_recuurence(months, offspring):
    parent, child = 1, 1
    for i in range(months-1):
        child, parent = parent, (parent + child * offspring)
    return child

# child o - They take one month to mature and then reproduce to produce n offspring.
# Adult/Parent 0 - They can reproduce to produce children, but recently matured produce offspring in the next cycle.

# 2 offspring produced in each cycle. 
# Month_1 = [o]
# Month_2 =  [0]
# Month_3 =  [0 o o]
# Month_4 =  [0 o o 0 0]
# Month_5 =  [0 o o 0 0 0 o o 0 o o]
```

```python
def rabbits_recuurence(months, offspring, life_span):
    parent, child = 1, 1
    for i in range(months-1):
        if i == (life_span + 1):
            parent = parent - 1
        elif i % 2 == 0 and i != 2 and i != 0:
            parent = parent - 1
        child, parent = parent, (parent + child * offspring)
    return child
```

```python
import re
Nucleotides = ["A","T","C","G"]
def clean_fasta_seq(file):
    """
    Cleans the FASTA sequence by:
    - Removing header lines (starting with '>')
    - Extracting sequence IDs and their corresponding sequences into a dictionary as key value pairs.
    - Removing whitespace and newline characters from the sequences.
    """

    match_id = re.findall(r">(.*?)(?=\W|$)",file,re.DOTALL)
    match_seq = re.findall(r"\n(.*?)(?=>|$)",file,re.DOTALL)

    # conventionally the . matches all character except the newline
    # re.DOTALL enables selection of all the characters.

    clean_dict = {}
    for id, seq in zip(match_id,match_seq):
        seq = seq.replace("\n","").upper()
        for i in seq: 
            if i not in Nucleotides:
                return False 
            else :
                clean_dict[id] = seq

    return clean_dict
```

```python
def min_len(file_name):
    max_len = 0
    min_len = float("inf")
    with open(file_name, "r") as file:
        for line in file:
            clean = line.strip()
            if clean.startswith(">") or not clean:
                pass
            length = len(clean)
            if length > max_len:
                max_len = length
            if length < min_len:
                min_len = length
                short_seq = clean
    return min_len,short_seq
```

```python
import pandas as pd
ids, seqs = [], []
current = ""
with open("file_rose.txt") as f:
    for line in f:
        line = line.strip()
        if line.startswith(">"):
            if current:
                seqs.append(current)
                current = ""
            ids.append(line[1:])
        else:
            current += line
    if current:
        seqs.append(current)
```

```python
def get_all_substrings(string):
    # Generates slices from every start index i to every end index j
    return [string[i:j] for i in range(len(string)) for j in range(i + 1, len(string) + 1)]

shortest = min(seqs, key = len)
str_list = set(get_all_substrings(shortest))
str_list = sorted(str_list,key = len, reverse = True)
```

```python
for length in range(len(shortest), 0, -1):

    seen = set()

    for start in range(len(shortest) - length + 1):

        sub = shortest[start:start+length]

        if sub in seen:
            continue
        seen.add(sub)

        if all(sub in seq for seq in seqs):
            print(sub)
            raise SystemExit("Longest Seq found Breaking the loop.")
```

## RNA Splicing

```python
# This is the more exact biologically
from Bio.Seq import Seq
import re
Seqs = ""
with open("file_rose.txt", "r") as file:
    for line in file:
        Seqs+=line
match_seq = re.findall(r"\n(.*?)(?=>|$)",Seqs,re.DOTALL)
cleaned_list = [item.strip() for item in match_seq]
raw_seq = cleaned_list.pop(0)
for i in cleaned_list:
    raw_seq = raw_seq.replace(i, "")
raw_seq = raw_seq.replace("\n","")
dna_seq = Seq(raw_seq)
print(dna_seq.translate())
```

```python
import requests
import re
for id in access_IDs:
    id_parse = id.split("_")
    response = requests.get(f"http://www.uniprot.org/uniprot/{id_parse[0]}\n.fasta")
    data = response.json()
    text = data["sequence"]["value"]
    pattern = r"(?=(N[^P][ST][^P]))"
    loc = []
    for match in re.finditer(pattern, text):
            loc.append(match.start()+1)
    if loc:
        print(id)
        print(*loc)
```
