```python
def fibonacci_no(position):
    old = new = 1
    for i in range(position-1):
        new, old = old, new+old
    return new
```

```python
def rabbits_recuurence(months, offspring):
    parent, child = 1, 1
    for i in range(months-1):
        child, parent = parent, (parent + child * offspring)
    return child

# child o - They take one month to mature and then reproduce to produce n offspring.
# Adult/Parent 0 - They can reproduce to produce children, but recently matured produce offspring in the next cycle.

# 2 offspring produced in each cycle. 
# Month_1 = [o]
# Month_2 =  [0]
# Month_3 =  [0 o o]
# Month_4 =  [0 o o 0 0]
# Month_5 =  [0 o o 0 0 0 o o 0 o o]
```

```python
# Let the total population be T, and the individual groups be m, n and k,
# with allelic composition as tall homo, tall hetero and dwarf homo, if we assume tallness as a factor (character) here

# Since it would be quite long procedure to calculate dominant probability we would find recessive population and minus it by 1 to get our answer

# So how can we get recessive offsprings, let's find out -:
# 1. We can mate two homozygous recessive parent.
# 2. We can mate hetrozygous father and hmozygous recessive female, and vice versa with gender. So will have two respective possibilities
# 3. We can mate two hetrerozygous parents.

# Since for both recessive homo the probability of bein recessive is 100%, but for the homo and hetero we would have to consider the cross
# taking place, this type of cross would produce 2 tall hetero plant and two dwarf plant so the prob becomes 1/2.
# Similarly for both hetero we have ratio of 3:1, 3 being dominant and 1 being recessive so here prob becomes 1/4.

# Now to correspond this with the variables given -:
# Probability of selecting each pair -:
# aa X aa - m/T * m-1/T-1 (since all are recessive no need for any extra prob multiply)
# Aa X Aa - n/T * n-1/T-1 * 1/4 (same as above but here we would have to multiple the prob we would out earlier 1/4)
# Aa X aa - n/T * m/T-1 * 2 * 1/2 (As the order of selection here can be of two types, first dwarf then hetero and vice versa. Again multiply by
# the recesssive prob we found).

# Now to get total recessive prob we would just find sum of these equations.

# P(A), A being the event = m*(m-1)\T*(T-1) + n*(n-1)/4T*(T-1) + n*m/T*(T-1)

# But the values here for m, n, k and T given in question to find P(A).

# P(domiant offsping) = 1 = P(A).

# Let's Check this -:
```

```python
import re
Nucleotides = ["A","T","C","G"]
def clean_fasta_seq(file):
    """
    Cleans the FASTA sequence by:
    - Removing header lines (starting with '>')
    - Extracting sequence IDs and their corresponding sequences into a dictionary as key value pairs.
    - Removing whitespace and newline characters from the sequences.
    """

    match_id = re.findall(r">(.*?)(?=\W|$)",file,re.DOTALL)
    match_seq = re.findall(r"\n(.*?)(?=>|$)",file,re.DOTALL)

    # conventionally the . matches all character except the newline
    # re.DOTALL enables selection of all the characters.

    clean_dict = {}
    for id, seq in zip(match_id,match_seq):
        seq = seq.replace("\n","").upper()
        for i in seq: 
            if i not in Nucleotides:
                return False 
            else :
                clean_dict[id] = seq

    return clean_dict
```
